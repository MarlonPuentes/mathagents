# Análisis de resultados
Ejecutar desde la raíz del repo después de las corridas reales. No escribe números a mano: todo sale de `results/*.jsonl`.

```
import glob, json, sys
sys.path.insert(0, '..')
from experiments.analyze import load_results, summarize, paired_comparisons
by_config = load_results(sorted(glob.glob('../results/*_test_*.jsonl')))
summary = [summarize(c, rows) for c, rows in sorted(by_config.items())]
summary
```

```
import matplotlib.pyplot as plt
rows = [s for s in summary if 'pass1' in s]
names = [s['configuration'] for s in rows]
p = [s['pass1'] for s in rows]
err = [[s['pass1'] - s['pass1_ci95'][0] for s in rows], [s['pass1_ci95'][1] - s['pass1'] for s in rows]]
plt.bar(names, p, yerr=err, capsize=4)
plt.ylabel('pass@1 (IC 95% Wilson)')
plt.savefig('../results/figura_pass1.png', dpi=200, bbox_inches='tight')
```
